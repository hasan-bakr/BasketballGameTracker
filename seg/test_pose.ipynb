```python
from ultralytics import YOLO
import torch
import cv2 as cv
```

```python
DEVICE = torch.device("cuda" if torch.cuda.is_available else "cpu") 
model = YOLO("yolo11m").to(DEVICE)
```

```python
import cv2 as cv

cap = cv.VideoCapture("videos/input/court.mp4")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    
    outputs = model(frame, conf=0.6)
    
    cv.imshow('Webcam Feed', outputs[0].plot())

    # Press 'q' to exit
    if cv.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv.destroyAllWindows()
```
